```python
import selfies as sf
import random
```

```python
alphabet = list(sf.get_semantic_robust_alphabet())

def random_selfies(length=12):
    return ''.join(random.choice(alphabet) for _ in range(length))

def generate_smiles():
    s = random_selfies()
    return sf.decoder(s)
```
