```python
def fitness(smiles):
    mol = Chem.MolFromSmiles(smiles)
    
    mw = Descriptors.MolWt(mol)
    donors = Descriptors.NumHDonors(mol)
    acceptors = Descriptors.NumHAcceptors(mol)

    # score simple (exemple pédagogique)
    score = -abs(mw - 300) - donors - acceptors

    return score
```

```python
import random

population = [
    "CCO",
    "CC(=O)O",
    "CCCC",
    "c1ccccc1",
    "CCN"
]
```

```python
def best_molecule(population):
    return max(population, key=fitness)
```

```python
def mutate(smiles):
    chars = list(smiles)

    if len(chars) == 0:
        return "C"

    i = random.randint(0, len(chars) - 1)
    chars[i] = random.choice(["C", "O", "N"])

    return "".join(chars)
```

```python
def optimize(iterations=50):

    pop = population.copy()

    for i in range(iterations):

        pop = sorted(pop, key=fitness, reverse=True)

        best = pop[:3]

        new_pop = best.copy()

        for mol in best:
            new_pop.append(mutate(mol))

        pop = new_pop

    return best_molecule(pop)
```

```python
result = optimize()

print("Meilleure molécule :", result)
print("Score :", fitness(result))
```
